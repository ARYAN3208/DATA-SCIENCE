```python
import numpy as np
```

```python
marks = [45, 32, 56, 75, 89, 54, 32, 89, 90, 87, 67, 54, 45, 98, 99, 67, 74]
minimum,Q1,median,Q3,maximum=np.quantile(marks, [0, 0.25, 0.5, 0.75, 1.0])
```

```python
minimum,Q1,median,Q3,maximum
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lower_fence=Q1-1.5*(IQR)
higer_fence=Q3+1.5*(IQR)
print(lower_fence,higer_fence)
```

```python
## anything less than lower fence and greater than higher fence will be considered as outlier
```

```python
import seaborn as sns
sns.boxplot(marks)
```

```python
## adding outliers 
marks_with_outliers=marks+[-100,-200,170]
sns.boxplot(marks_with_outliers)
```
