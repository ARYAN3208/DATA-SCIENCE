ALGERIAN FOREST FIRES DATASET
The dataset used is the Algerian Forest Fires dataset, which contains data from two regions: Bejaia and Sidi Bel Abbes. We will combine these datasets for our analysis. For classification problems, the output feature can be the class label, while for regression problems, the dependent feature will be the Fire Weather Index (FWI), and the remaining features will be independent variables.

The dataset includes 244 instances from two regions of Algeria: Bejaia, located in the northeast, and Sidi Bel Abbes, located in the northwest. The data covers the period from June 20th, 2012, to September 2012. It contains 11 attributes and one class output attribute. The instances are classified into fire and non-fire classes, allowing use for classification tasks as well.

Attributes include date, temperature (in Celsius), relative humidity (RH), wind speed (WS), rain (total daily), fine fuel moisture code (FMC), drought code (DC), and other indices from the Fire Weather Index system.

Attribute Information (Printable Format)

Date: (DD/MM/YYYY) Day, month (June to September), year (2012). Weather data observations.

Temp: Temperature at noon (maximum temperature) in Celsius degrees: 22 to 42.

RH: Relative Humidity in %: 21 to 90.

Ws: Wind speed in km/h: 6 to 29.

Rain: Total rainfall in mm: 0 to 16.8.

FWI Components

Fine Fuel Moisture Code (FFMC) index from the FWI system: 28.6 to 92.5.

Duff Moisture Code (DMC) index from the FWI system: 1.1 to 65.9.

Drought Code (DC) index from the FWI system: 7 to 220.4.

Initial Spread Index (ISI) index from the FWI system: 0 to 18.5.

Buildup Index (BUI) index from the FWI system: 1.1 to 68.

Fire Weather Index (FWI) index: 0 to 31.1.

Target Class

Classes: Two classes — Fire and Not Fire.
