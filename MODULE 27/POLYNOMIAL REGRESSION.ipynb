```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Creating the dataset
np.random.seed(0)  # For reproducibility
x = 6* np.random.rand(100, 1) - 3  # Random values shifted by -3

# Quadratic relationship with noise
# y = 0.5 * x^2 + 1.5 * x + 2 + noise
noise = np.random.randn(100, 1)
y= 0.5 * x**2 + 1.5 * x+ 2 + noise
plt.scatter(x,y,color='blue')
plt.xlabel("x dataset")
plt.ylabel("y dataset")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
### lets implement a simple linear regression
from sklearn.linear_model import LinearRegression
regression=LinearRegression()
```

```python
x_train
```

```python
y_train
```

```python
regression.fit(x_train,y_train)
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,regression.predict(x_test))
print(score)
```

```python
### lets visulaize 
plt.plot(x_train,regression.predict(x_train))
plt.scatter(x_train,y_train)
plt.xlabel("x dataset")
plt.ylabel("y dataset")
plt.show()
```

```python
### lets apply polynomial transfomration
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly=PolynomialFeatures(degree=2,include_bias=True)
x_train_poly=poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)
```

```python
x_train_poly
```

```python
x_test_poly
```

```python
from sklearn.metrics import r2_score
regression=LinearRegression()
regression.fit(x_train_poly,y_train)
y_pred=regression.predict(x_test_poly)
score=r2_score(y_test,y_pred)
print(score)
```

```python
print(regression.coef_)
```

```python
print(regression.intercept_)
```

```python
plt.scatter(x_train,regression.predict(x_train_poly))
plt.scatter(x_train,y_train)
plt.show()
```

```python
poly=PolynomialFeatures(degree=3,include_bias=True)
x_train_poly=poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)
```

```python
x_train_poly
```

```python
from sklearn.metrics import r2_score
regression=LinearRegression()
regression.fit(x_train_poly,y_train)
y_pred=regression.predict(x_test_poly)
score=r2_score(y_test,y_pred)
print(score)
```

```python
## prediction of new dataset
x_new = np.linspace(-3,3,200).reshape(200,1)
x_new_poly = poly.transform(x_new)
x_new_poly
```

```python
# Plotting predictions for new data along with training and testing points
y_new = regression.predict(x_new_poly)
plt.scatter(x_train, y_train, color='green', label='Training Data')
plt.scatter(x_test, y_test, color='blue', label='Testing Data')
plt.plot(x_new, y_new, color='red', label='Polynomial Regression Prediction')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Polynomial Regression Predictions on New Data')
plt.legend()
plt.show()
```

```python
from sklearn.pipeline import Pipeline

def poly_regression(degree):
    X_new = np.linspace(-3, 3, 200).reshape(200, 1)

    poly_features = PolynomialFeatures(degree=degree, include_bias=True)
    lin_reg = LinearRegression()

    poly_regression = Pipeline([
        ("poly_features", poly_features),
        ("lin_reg", lin_reg)
    ])

    poly_regression.fit(x_train, y_train)
    y_pred_new = poly_regression.predict(X_new)

    # Plotting prediction line
    plt.plot(x_new, y_pred_new, "r", label="Degree " + str(degree), linewidth=3)
    plt.plot(x_train, y_train, "b.", linewidth=3)
    plt.plot(x_test, y_test, "g.", linewidth=3)
    plt.legend(loc="upper left")

    plt.xlabel("x")
    plt.ylabel("y")
    plt.axis([-4, 4, 0, 10])
```

```python
poly_regression(5)
plt.show()
```

```python
poly_regression(10)
plt.show()
```
