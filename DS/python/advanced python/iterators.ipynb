```python
##Iterators are an advanced Python concept that allows for efficient looping and memory management.
## Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.
```

```python
my_list = [1,2,3,45,6,]
```

```python
print(my_list)
```

```python
type(my_list)
```

```python
### creating an iterator
iterator=iter(my_list) 
print(iterator)
```

```python
### iterate through all elements
next(iterator)
```

```python
iterator = iter(my_list)

try :
   next(iterator)
except StopIteration :
   print("there is no element in the iterator") 
```

```python
 ## string iterator
 my_string = 'he'
string_iterator = iter(my_string)
print(next(string_iterator))
print(next(string_iterator))
```
