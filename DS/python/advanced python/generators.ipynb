```python
### Generators provide a simple way to create iterators by using the yield keyword to produce a series of values lazily. This means they generate values on the fly and do not store them in memory, enabling efficient memory management during looping.
```

```python
def square(n):
    for i in range(3):
        return i**2
```

```python
square(3)
```

```python
def square(n):
    for i in range(3):
        yield i**2
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
a  = square(3)
a
```

```python
next(a)
```

```python
next(a)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen = my_generator()
gen
```

```python
next(gen)
```

```python
next(gen)
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```

```python
#### reading large files VERY PRACTICAL EXAMPLES
###Generators are particularly useful for reading large files because they allow processing one line at a time without loading the entire file into memory. This efficient approach helps manage memory usage when dealing with large datasets.
```

```python
def read_large_files(file_path):
    with open(file_path,'r') as file:
        for line in file: 
          yield line
```

```python
 file_path  = 'large_file.txt'
 for line in read_large_files(file_path):
       print(line)
```

```python
##The function read_large_file opens a file and yields one line at a time. This allows the caller to process each line lazily without loading the entire file into memory.
```
