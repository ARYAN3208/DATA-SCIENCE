```python
###A pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to a column in a table. The basic difference between a Series and a DataFrame is that a Series is one-dimensional, while a DataFrame is two-dimensional, mutable, and can contain heterogeneous data.
```

```python
###### series - 1d array 
import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)  ### Series should be capital according to indentation
print(series)
print(type(series))
```

```python
### create a series from dictionary
data_dict = {'a': 1, 'b': 2, 'c': 3}
series_dict = pd.Series(data_dict)
print(series_dict)
```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data,index)
```

```python
### dataframe -  2d array like object
## create a dataframe from dictionary
data = {
    'name': ['Krish', 'John', 'Jack'],
    'age': [25, 30, 45],
    'city': ['Bangalore', 'New York', 'Florida']
}
df = pd.DataFrame(data)
print(df)
```

```python
import numpy as np
np.array(df)
```

```python
## creating a data frame from list of dictionaries
data = [
    {'name': 'Krish', 'age': 32, 'city': 'Bangalore'},
    {'name': 'John', 'age': 34, 'city': 'Bangalore'},
    {'name': 'Jack', 'age': 32, 'city': 'Bangalore'}
]
df = pd.DataFrame(data)
print(df)
```
