```python
from nltk.stem import WordNetLemmatizer
```

```python
import nltk
nltk.download('wordnet')
lemmatizer=WordNetLemmatizer()
```

Pos = 'n' means the roort word will come as noun and if pos = 'v' then root word become a verb

```python
## NOUN - n
## verb - v
## adjective - a 
## adverb - r
lemmatizer.lemmatize("going",pos='v')
```

```python
words=['eating','eaten','eats','going','gone','goes']
```

```python
for word in words:
    print(word + " -----> " + lemmatizer.lemmatize(word,pos="n"))
```
