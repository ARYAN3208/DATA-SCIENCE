```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
corpus="""hello welcome to my world of aiml.
please do come ! and visit."""
```

```python
print(corpus)
```

```python
### tokenization
### convert sentence to paragraph
from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
```

punkt: The main trained model that detects where sentences usually end.

punkt_tab: The rule tables (like abbreviations and punctuation rules) that make sentence detection accurate.

```python
### convert paragraph to words or sentence to words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

With the help os this wordpunct even full stop, ! etc also gets separated

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokens=TreebankWordTokenizer()
```

```python
tokens.tokenize(corpus)
```

Here for only last word full stop is separate and for rest sentence the full stop is attached.
