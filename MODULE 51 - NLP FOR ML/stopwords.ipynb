```python
paragraph="""Dr. A. P. J. Abdul Kalam was a visionary scientist and an inspiring leader who dedicated his life to the development of India and the empowerment of its youth.
 Known as the “Missile Man of India,” he played a key role in strengthening the country’s space and defense programs and contributed significantly to projects like the SLV-III and the Agni and Prithvi missiles.
 As the 11th President of India, he was admired for his simplicity, humility, and deep connection with students.
 His teachings emphasized dreaming big, hard work, discipline, and the importance of education in building a strong nation.
 Dr. Kalam’s life and words continue to motivate millions to pursue excellence and serve the country with integrity and passion.
"""
```

```python
stemmer=PorterStemmer()
```

```python
from nltk.tokenize import sent_tokenize
sentences=nltk.sent_tokenize(paragraph)
```

```python
### apply stopwords and filter and then apply stemming
for i in range(len(sentences)):
  words=nltk.word_tokenize(sentences[i])
  words =[stemmer.stem(word)for word in words if word not in set(stopwords.words('english'))]
  sentences[i]=' '.join(words)
```

```python
import nltk
from nltk.stem import SnowballStemmer
from nltk.corpus import stopwords

nltk.download('punkt')
nltk.download('stopwords')

stemmer = SnowballStemmer("english")   # Snowball stemmer object
stop_words = set(stopwords.words('english'))

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [stemmer.stem(word) for word in words if word.lower() not in stop_words]
    sentences[i] = ' '.join(words)
```

```python
sentences
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
for i in range (len(sentences)):
  words=nltk.word_tokenize(sentences[i])
  words =[lemmatizer.lemmatize(word) for word in words if word.lower() not in stop_words]
  sentences[i]=' '.join(words)
```

CC is a coordinating conjunction like and, but, or. CD is a cardinal number like one, two, 5. DT is a determiner like a, an, the. EX is existential there as in there is. FW is a foreign word like bonjour. IN is a preposition or subordinating conjunction like in, on, at, because. JJ is an adjective like big. JJR is a comparative adjective like bigger. JJS is a superlative adjective like biggest. LS is a list marker like 1), A). MD is a modal verb like can, could, will, must. NN is a singular or mass noun like desk, water. NNS is a plural noun like desks. NNP is a singular proper noun like Harrison, India. NNPS is a plural proper noun like Americans, Indians. PDT is a predeterminer like all, both, half. POS is a possessive ending like ’s. PRP is a personal pronoun like I, you, he, she, they. PRP$ is a possessive pronoun like my, your, his, her, their. RB is an adverb like quickly, very. RBR is a comparative adverb like better, faster. RBS is a superlative adverb like best, fastest. RP is a particle like up, off, out in give up. TO is the word to used before verbs like to go, to eat. UH is an interjection like oh, wow, hey. VB is the base form of a verb like go, eat. VBD is the past tense verb like went, ate. VBG is the gerund or present participle like going, eating. VBN is the past participle like gone, eaten. VBP is a verb in present tense not in third-person singular like go, eat. VBZ is a verb in third-person singular present like goes, eats. WDT is a wh-determiner like which, that. WP is a wh-pronoun like who, what. WP$ is a possessive wh-pronoun like whose. WRB is a wh-adverb like where, when, why.
